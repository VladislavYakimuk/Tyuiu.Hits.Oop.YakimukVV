Класс Растение

```polyglot-notebook
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public virtual void Bloom() 
    {
        Console.WriteLine($"Растение {Name} ({Type}) цветет ");
    }

    public void Grow(float amount) 
    {
        Height += amount;
        Console.WriteLine($"Растение {Name} вырос на {amount} см. Текущая высота: {Height} см.");
    }
}

var plant = new Plant { Name = "Роза", Type = "Цветок", Height = 30.0f };
plant.Grow(5.0f);
plant.Bloom();
```

Класс Цветок

```polyglot-notebook
public class Flower : Plant
{
    public string Color { get; set; }
    public string Count { get; set; }

    public override void Bloom() 
    {
        Console.WriteLine($"Цветок {Name} ({Type}) с цветом {Color} цветет");
    }

    public void Smell() 
    {
        Console.WriteLine($"Цветок имеет лепестков {Count}. Он увядает");
    }
}

var flower = new Flower { Name = "Тюльпан", Type = "Цветок", Height = 25.0f, Color = "красный", Count = "6" };
flower.Bloom();
flower.Smell();
```

Класс дерево

```polyglot-notebook
public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public override void Bloom() 
    {
        Console.WriteLine($"{Name} ({Type}) с типом листьев {LeafType} цветет");
    }

    public void AgeUp(int amount) 
    {
        Age += amount;
        Console.WriteLine($"Дерево {Name} стало старше на {amount} лет . Текущий возраст: {Age} лет.");
    }
}

var tree = new Tree { Name = "Дуб", Type = "Дерево", Age = 50, LeafType = "лиственные" };
tree.AgeUp(5);
tree.Bloom();

```

Полиморфизм

```polyglot-notebook
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public virtual void Bloom()
    {
        Console.WriteLine($"{Name} ({Type}) цветет.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }
}

Plant myFlower = new Flower { Name = "Лилия", Type = "Цветок", Height = 40.0f };
myFlower.Bloom();
Plant myTree = new Tree { Name = "Сосна", Type = "Дерево", Height = 300.0f };
myTree.Bloom();
```
